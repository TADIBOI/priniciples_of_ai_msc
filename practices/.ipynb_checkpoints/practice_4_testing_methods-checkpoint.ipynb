```python
State = tuple[int, ...]

goal: State = (1, 2, 3, 8, 0, 4, 7, 6, 5)
```

```python
def misplaced(state: State) -> int:
# how many items are misplaced 
    n = 0
    for i in range(len(goal)):
        if state[i] != goal[i]:
            n += 1 
    
    return n
```

```python
print(misplaced((1,3,2,4,5,6,7,8,0)))
```
